# Outlier Analysis

## Objective

Identify, analyze, and handle potential outliers in the Titanic dataset.

```python
import pandas as pd
import numpy as np
```

## 1. Load the Dataset

Load the Titanic dataset into a Pandas DataFrame.

```python
df = pd.read_csv("/Volumes/workspace/default/ml_data/Titanic-Dataset.csv")
```

## 2. Understand Outliers

Identify values that are unusually far from most other values in a dataset.

An outlier is not automatically an error and should be investigated before removal or modification.

## 3. Calculate Q1 and Q3

Calculate the first quartile (Q1) and third quartile (Q3) of passenger ages.

```python
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)

print("Q1:", Q1)
print("Q3:", Q3)
```

## 4. Interquartile Range (IQR)
### FORMULA
IQR = Q3 - Q1


```python
IQR = Q3 - Q1

print("IQR:", IQR)
```

## 5. Calculate Outlier Boundaries

Calculate the lower and upper boundaries using the IQR method.

###Formula
Lower Bound = Q1 - 1.5 × IQR, 
Upper Bound = Q3 + 1.5 × IQR

```python
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Lower Bound:", lower)
print("Upper Bound:", upper)
```

## 6. Identify Potential Outliers

Identify passenger ages that fall outside the IQR boundaries.

```python
outliers = df[(df["Age"]>upper) | (df["Age"] < lower)]
outliers[["Name", "Age"]]
```

## 7. Inspect Potential Outliers

Inspect the passenger details associated with the potential outliers.

```python
outliers[["Name","Age","Sex","Pclass","Fare"]]
```

## 8. Evaluate Potential Outliers

Determine whether the identified outliers represent valid observations or possible data errors.

Valid unusual values should be retained unless there is evidence that they are incorrect.

### Observation

The identified older ages are plausible passenger ages rather than obvious data errors. Retain these observations for further analysis.

## 9. Remove Outliers

Remove observations that fall outside the IQR boundaries when there is a valid reason to exclude them.

```python
clean_data = df[(df["Age"] >= lower) & (df["Age"] <= upper)]

clean_data[["Name", "Age"]]
```

## 10. Replace Outliers

Replace outlier values with the median when there is a valid reason to reduce the effect of extreme values.

```python
median_age = df["Age"].median()

replaced_data = df.copy()
replaced_data.loc[replaced_data["Age"] > upper, "Age"] = median_age

replaced_data[["Name", "Age"]]
```

## 11. Cap Outliers

Cap values outside the IQR boundaries at the corresponding lower or upper boundary.

```python
capped_data = df["Age"].clip(lower=lower, upper=upper)

capped_data
```

## 12. Transform the Data

Apply a logarithmic transformation to reduce the influence of large values.
np.log1p(x) calculates log(1 + x) and safely handles zero values.

```python
log_data = np.log1p(df["Age"])

log_data
```
