# Data Visualization

## Objective

Visualize the Titanic dataset to identify patterns, comparisons, distributions, and relationships between variables.

```python
import pandas as pd
```

## 1. Load the Dataset

Load the Titanic dataset into a Pandas DataFrame.

```python
df = pd.read_csv("/Volumes/workspace/default/ml_data/Titanic-Dataset.csv")
```

## 2. Line Plot

Visualize passenger fares across passenger IDs using a line plot.

```python
df.plot(x="PassengerId", y="Fare", kind="line")
```

## 3. Bar Plot

Compare the number of passengers in each `Sex` category using a bar plot.

```python
df["Sex"].value_counts().plot(kind="bar")
```

## 4. Bar Plot for Embarkation Points

Compare the number of passengers from each `Embarked` category using a bar plot.

```python
df["Embarked"].value_counts().plot(kind="bar")
```

## 5. Pie Chart

Visualize the proportion of passengers in each `Sex` category using a pie chart.

```python
df["Sex"].value_counts().plot(
    kind = "pie",
    autopct = "%1.1f%%"
    )
```

## 6. Box Plot

Visualize the distribution of passenger ages and identify potential outliers using a box plot.

```python
df.boxplot(column="Age")
```

## 7. Histogram

Visualize the distribution of passenger ages using a histogram.

```python
df["Age"].hist(bins=20)
```

## 8. Scatter Plot

Visualize the relationship between passenger age and fare using a scatter plot.

```python
df.plot.scatter(x="Age", y = "Fare")
```

## 9. Visualization Summary

Use appropriate plots to understand different aspects of the dataset.

- Use line plots to visualize ordered numerical values.
- Use bar plots to compare categorical values.
- Use pie charts to visualize category proportions.
- Use box plots to examine numerical distributions and potential outliers.
- Use histograms to understand the distribution of numerical values.
- Use scatter plots to examine relationships between two numerical variables.
