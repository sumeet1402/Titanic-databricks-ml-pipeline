# 11 — Model Evaluation

## Objective

Evaluate the Logistic Regression model using accuracy, confusion matrix, precision, recall, F1-score, and a classification report.

```python
df = spark.table("default.titanic_encoded").toPandas()
df.shape
```

```python
X = df.drop(columns= ["PassengerId", "Name", "Ticket", "Cabin", "Survived"])
y = df["Survived"]

X.shape, y.shape
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, 
    y, 
    test_size = 0.20,
    random_state= 42
)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)

model.fit(X_train, y_train)
```

```python
y_pred  = model.predict(X_test)

y_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test, y_pred)
accuracy
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
cm
```

```python
from sklearn.metrics import precision_score
precision = precision_score(y_test, y_pred)
precision
```

```python
from sklearn.metrics import recall_score
recall = recall_score(y_test, y_pred)
recall
```

```python
from sklearn.metrics import f1_score
f1 = f1_score(y_test, y_pred)
f1
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))

```

## Results

The baseline model achieved approximately **82.68% accuracy**, **78.67% precision**, **79.73% recall**, and **79.19% F1-score** on the held-out test set.

The confusion matrix was `[[89, 16], [15, 59]]`.
