# 10 — Model Training

## Objective

Train a Logistic Regression classifier and inspect its learned coefficients and prediction probabilities.

```python
df = spark.table("default.titanic_encoded").toPandas()
```

```python
df.shape
```

```python
X = df.drop(columns=["PassengerId", "Name", "Ticket", "Cabin", "Survived"])
y = df["Survived"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(max_iter=1000)
```

```python
model.fit(X_train, y_train)
```

```python
import pandas as pd

coefficients = pd.DataFrame({
    "Feature" : X_train.columns,
    "Coefficient" : model.coef_[0]
    })

coefficients
```

```python
model.predict(X_test)
```

```python
model.predict_proba(X_test)[:5]
```

## Conclusion

A Logistic Regression model was trained on the training split and used to generate test predictions and class probabilities. Evaluation is performed in the next notebook.
