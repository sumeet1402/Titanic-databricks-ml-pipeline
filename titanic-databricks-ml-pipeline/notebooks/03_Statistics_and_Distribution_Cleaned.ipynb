# Statistics and Distribution

## Objective

Calculate statistical measures and understand the distribution of numerical data.

```python
import pandas as pd
import numpy as np
```

## 1. Load the Dataset

Load the Titanic dataset into a Pandas DataFrame.

```python
df = pd.read_csv("/Volumes/workspace/default/ml_data/Titanic-Dataset.csv")
```

## 2. Calculate the Mean

Calculate the average age of the passengers.

```python
df["Age"].mean()
```

## 3. Calculate the Median

Calculate the median age of the passengers.

```python
df["Age"].median()
```

## 4. Calculate the Mode

Calculate the most frequently occurring age.

```python
df["Age"].mode()
```

## 5. Calculate the Variance

Calculate the variance of passenger ages.

```python
df["Age"].var()
```

## 6. Calculate the Standard Deviation

Calculate the standard deviation of passenger ages.

```python
df["Age"].std()
```

## 7. Generate a Statistical Summary

Generate a statistical summary of the numerical columns.


```python
df.describe()
```

## 8. Calculate Skewness

Calculate the skewness of passenger ages.

```python
df["Age"].skew()
```

## 9. Calculate Kurtosis

Calculate the kurtosis of passenger ages.

```python
df["Age"].kurt()
```

## 10. Examine the Age Distribution

Examine how passenger ages are distributed across different age ranges.

```python
df["Age"].hist(bins=20)
```

## 11. Examine the Fare Distribution

Examine how passenger fares are distributed across different fare ranges.

```python
df["Fare"].hist(bins=20)
```

## 12. Identify the Distribution Shape

Identify whether the Age and Fare distributions are approximately symmetric or skewed.

```python
print("Age skewness:", df["Age"].skew())
print("Fare skewness:", df["Fare"].skew())
```

## Conclusion

Summarize the statistical characteristics and distribution of the numerical data.

- Calculate mean, median, mode, variance, and standard deviation.
- Use `describe()` to obtain a statistical summary.
- Use skewness and kurtosis to understand distribution shape and tail behavior.
- Examine numerical distributions using histograms.
- Identify patterns, spread, and potential extreme values before further analysis.
